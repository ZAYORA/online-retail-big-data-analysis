```python
# Import Library
from pyspark.sql import SparkSession
from pyspark.sql.functions import col
import pandas as pd
```

```python
# Membuat Spark Session
spark = SparkSession.builder \
    .appName("Analisis Online Retail Big Data Tugas Individu 2") \
    .getOrCreate()
```

```python
# Membaca file Excel
pdf = pd.read_excel("Online Retail.xlsx")

pdf.head()
```

```python
# Mengubah pandas dataframe menjadi pyspark dataframe
df = spark.createDataFrame(pdf)

# Menampilkan data
df.show(5)
```

```python
# Menampilkan struktur dataset
df.printSchema()
```

```python
# Menampilkan jumlah baris data
print("Jumlah Data:", df.count())
```

```python
# Menampilkan beberapa kolom penting
df.select(
    "InvoiceNo",
    "Description",
    "Quantity",
    "UnitPrice",
    "Country"
).show(5)
```

```python
# Menambahkan kolom TotalPrice
df2 = df.withColumn(
    "TotalPrice",
    col("Quantity") * col("UnitPrice")
)

df2.show(5)
```

```python
# Filter data Quantity lebih dari 20
df_filter = df2.filter(
    col("Quantity") > 20
)

df_filter.show(10)
```

```python
# Filter data transaksi dari France
df_france = df2.filter(
    col("Country") == "France"
)

df_france.show(10)
```

```python
# Group By Country dan menghitung total penjualan
df2.groupBy("Country") \
   .sum("TotalPrice") \
   .show()
```

```python
# Group By produk dan menghitung total quantity
df2.groupBy("Description") \
   .sum("Quantity") \
   .show(10)
```

```python
# Mengurutkan data berdasarkan TotalPrice terbesar
df2.orderBy(
    col("TotalPrice").desc()
).show(10)
```

```python
# Mengurutkan data berdasarkan Quantity terbesar
df2.orderBy(
    col("Quantity").desc()
).show(10)
```

```python
# Mengecek jumlah partition
print(df2.rdd.getNumPartitions())
```

```python
# Repartition data menjadi 4 partition
df_repartition = df2.repartition(4)

print(df_repartition.rdd.getNumPartitions())
```

```python
# Menampilkan statistik deskriptif
df2.describe().show()
```

```python
# Menampilkan jumlah transaksi tiap negara
df2.groupBy("Country") \
   .count() \
   .orderBy(col("count").desc()) \
   .show()
```

```python
# Menampilkan produk unik
df2.select("Description").distinct().show(10)
```

```python
# Menghapus data null
df_clean = df2.dropna()

df_clean.show(5)
```

```python
# Menyimpan hasil ke CSV
df_clean.write.csv(
    "output_online_retail",
    header=True,
    mode="overwrite"
)
```

Dari praktikum ini, PySpark dapat digunakan untuk membaca dan mengolah dataset berukuran besar. Beberapa operasi yang dilakukan antara lain filtering data, agregasi menggunakan groupBy, sorting data, serta pengelolaan partition pada Spark.

Berdasarkan hasil analisis, United Kingdom memiliki jumlah transaksi tertinggi pada dataset Online Retail. Selain itu, beberapa produk memiliki quantity penjualan yang jauh lebih besar dibanding produk lainnya.
